```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
```

```python
X,y=load_digits(return_X_y=True) # if true, returns (data,target)
```

```python
X.shape,y.shape
```

```python
plt.figure(figsize=(8,8))
plt.imshow(X[3].reshape(8,8),cmap='Greys')
```

```python
fig,axes=plt.subplots(nrows=2,ncols=3,sharex=True,sharey=True,figsize=(12,8))
axes[0,0].imshow(X[204].reshape(8,8),cmap='Greys')
axes[0,1].imshow(X[100].reshape(8,8),cmap='Greys')
axes[0,2].imshow(X[205].reshape(8,8),cmap='Greys')
axes[1,0].imshow(X[3].reshape(8,8),cmap='Greys')
axes[1,1].imshow(X[207].reshape(8,8),cmap='Greys')
axes[1,2].imshow(X[134].reshape(8,8),cmap='Greys')
plt.tight_layout()
```

#### Fisher Discrminant Analysis is also known as Linear Discriminant Analysis

```python
import pandas as pd
import seaborn as sns
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
lda=LinearDiscriminantAnalysis(n_components=2)

X_lda=lda.fit_transform(X,y=y)
data=pd.DataFrame({'FDA1':X_lda[:,0],'FDA2':X_lda[:,1],'class':y})

fig=plt.figure(figsize=(8,8))
sns.scatterplot(x='FDA1',y='FDA2',hue='class',data=data,palette='tab10')

```

```python
data.head()
```

#### 3-D

```python
lda=LinearDiscriminantAnalysis(n_components=3)
X_lda=lda.fit_transform(X,y=y)

data=pd.DataFrame({'FDA1':X_lda[:,0],'FDA2':X_lda[:,1],'FDA3':X_lda[:,2],'class':y})
```

```python
data.head()
```

```python
import plotly.express as px
fig=px.scatter_3d(data, x='FDA1',y='FDA2',z='FDA3',color='class')
fig.show()
```
